```python
%%writefile logic.py
import numpy as np
import random
import pandas as pd

# Global Settings
NUM_SQUARES = 100
ABSORBING = 100
ladders = {3: 22, 5: 8, 11: 26, 20: 29, 27: 56, 40: 59, 83: 99}
snakes = {17: 4, 19: 7, 21: 9, 54: 34, 62: 28, 84: 17, 97: 20}
BOARD_SPECIALS = {**ladders, **snakes}

class MultiplayerGame:
    def __init__(self, ladders, snakes, first_mover='A'):
        self.NUM_SQUARES = 100
        self.ABSORBING = 100
        self.board_specials = {**ladders, **snakes}
        self.player_A = 0
        self.player_B = 0
        self.current_turn = first_mover
        self.move_history = []

    def roll_dice(self):
        return random.randint(1, 6)

    def apply_move(self, position, roll):
        if position == self.ABSORBING: return self.ABSORBING
        next_pos = position + roll
        if next_pos >= self.NUM_SQUARES: return self.ABSORBING
        return self.board_specials.get(next_pos, next_pos)

    def play_game(self):
        turn_count = 0
        while turn_count < 1000: # Safety limit
            turn_count += 1
            roll = self.roll_dice()
            player_name = self.current_turn
            old_pos = self.player_A if player_name == 'A' else self.player_B

            if player_name == 'A':
                self.player_A = self.apply_move(self.player_A, roll)
                new_pos = self.player_A
            else:
                self.player_B = self.apply_move(self.player_B, roll)
                new_pos = self.player_B

            self.move_history.append({
                "Turn": turn_count, "Player": player_name,
                "Roll": roll, "From": old_pos, "To": new_pos
            })

            if new_pos == 100:
                return player_name, turn_count
            self.current_turn = 'B' if self.current_turn == 'A' else 'A'
        return 'None', turn_count

    def get_game_stats(self):
        return self.move_history

def run_monte_carlo_simulation(num_episodes):
    turns = []
    winners = []
    for _ in range(num_episodes):
        g = MultiplayerGame(ladders, snakes)
        winner, t = g.play_game()
        turns.append(t)
        winners.append(winner)
    return turns, winners

def build_transition_matrix(ladders, snakes):
    P = np.zeros((101, 101))
    for state in range(1, 101):
        if state == 100:
            P[state, state] = 1
            continue
        for roll in range(1, 7):
            next_s = state + roll
            if next_s > 100: next_s = 100
            else: next_s = BOARD_SPECIALS.get(next_s, next_s)
            P[state, next_s] += 1/6
    return P
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
from logic import MultiplayerGame, run_monte_carlo_simulation, build_transition_matrix, ladders, snakes
import plotly.express as px
import plotly.graph_objects as go

# ---------- Page Config ----------
st.set_page_config(
    layout="wide",
    page_title="Snake & Ladder Simulator 🎲",
    page_icon="🎲"
)

# ---------- Custom CSS ----------
st.markdown("""
<style>
    .stApp {background-color: #f8f9fa;}
    h1 {color: #1e3a8a;}
    .stMetric {background-color: #ffffff; padding: 15px; border-radius: 10px; box-shadow: 2px 2px 10px rgba(0,0,0,0.1);}
    div[data-testid="stExpander"] {border: none; box-shadow: none;}
</style>
""", unsafe_allow_html=True)

# ---------- Caching Function (Speed Boost) ----------
@st.cache_data
def cached_simulation(sims):
    """Run simulation and cache results so it doesn't re-run on every interaction"""
    turns, winners = run_monte_carlo_simulation(sims)
    return turns, winners

# ---------- Title ----------
st.title("🎲 Snake & Ladder: Probabilistic Analysis")
st.markdown("### A Data Science Approach to a Classic Game")

# ---------- Sidebar ----------
with st.sidebar:
    st.header("⚙️ Settings")
    sims = st.slider("Monte Carlo Simulations", 100, 5000, 1000, step=100)
    start_player = st.radio("Who starts first?", ["Player A", "Player B"])
    st.info("💡 **Tip:** Higher simulations give more accurate probabilities but take longer.")

# ---------- Tabs ----------
tab1, tab2, tab3 = st.tabs(["📊 Simulation Analytics", "🧮 Markov Chain Matrix", "🎮 Live Gameplay"])

# ---------- Tab 1: Simulation ----------
with tab1:
    st.header("Monte Carlo Analysis")

    if st.button("🚀 Run Simulation"):
        with st.spinner(f'Simulating {sims} games...'):
            turns, winners = cached_simulation(sims)

            # --- Metrics Row ---
            col1, col2, col3, col4 = st.columns(4)
            col1.metric("Avg Turns to Win", f"{np.mean(turns):.2f}")
            col2.metric("Min Turns (Luckiest)", f"{min(turns)}")
            col3.metric("Max Turns (Unluckiest)", f"{max(turns)}")
            col4.metric("Player A Win Rate", f"{(winners.count('A')/sims)*100:.1f}%")

            # --- Plots ---
            c1, c2 = st.columns([2, 1])

            with c1:
                # Distribution Plot
                fig_turns = px.histogram(turns, nbins=40,
                                       title=f"Distribution of Game Duration ({sims} Episodes)",
                                       labels={'value': 'Turns to Win'},
                                       color_discrete_sequence=['#3b82f6'])
                fig_turns.add_vline(x=np.mean(turns), line_dash="dash", line_color="red", annotation_text="Mean")
                st.plotly_chart(fig_turns, use_container_width=True)

            with c2:
                # Win/Loss Pie Chart
                df_wins = pd.DataFrame(winners, columns=['Winner'])
                fig_pie = px.pie(df_wins, names='Winner', title="Win Probability",
                               color='Winner', color_discrete_map={'A':'#3b82f6', 'B':'#ef4444'})
                st.plotly_chart(fig_pie, use_container_width=True)

            # --- Data Export (The Pro Feature) ---
            st.markdown("---")
            st.subheader("💾 Export Data")
            df_results = pd.DataFrame({'Game_ID': range(1, sims+1), 'Winner': winners, 'Turns': turns})
            csv = df_results.to_csv(index=False).encode('utf-8')

            st.download_button(
                label="📥 Download Results as CSV",
                data=csv,
                file_name='snake_ladder_simulation.csv',
                mime='text/csv',
            )

# ---------- Tab 2: Transition Matrix ----------
with tab2:
    st.header("Markov Transition Matrix")
    st.markdown("""
    This heatmap represents the probability $P_{ij}$ of moving from Square $i$ to Square $j$.
    * **Diagonal:** No movement (impossible in this version).
    * **Jumps:** Notice the distant yellow dots representing Snakes and Ladders.
    """)

    P = build_transition_matrix(ladders, snakes)
    fig = px.imshow(P, labels=dict(x="Target Square", y="Current Square", color="Prob"),
                    color_continuous_scale='Viridis', origin='lower')
    fig.update_layout(height=600)
    st.plotly_chart(fig, use_container_width=True)

# ---------- Tab 3: Live Game ----------
with tab3:
    st.header("🎮 Interactive Live Game")

    col_game, col_log = st.columns([1, 2])

    with col_game:
        st.write("Click to simulate a single match move-by-move.")
        if st.button("🎲 Play New Game"):
            game = MultiplayerGame(ladders, snakes, first_mover='A' if start_player == "Player A" else 'B')
            winner, total_turns = game.play_game()

            # Winner Announcement
            st.balloons()
            if winner == 'A':
                st.success(f"🏆 Player A Wins in {total_turns} turns!")
            else:
                st.error(f"🏆 Player B Wins in {total_turns} turns!")

            # Save history to session state to display in the other column
            st.session_state['history'] = game.get_game_stats()

    with col_log:
        if 'history' in st.session_state:
            df = pd.DataFrame(st.session_state['history'])

            # Progress Chart
            st.subheader("Race Progress")
            fig_race = px.line(df, x='Turn', y='To', color='Player', markers=True,
                             color_discrete_map={'A':'blue', 'B':'red'})
            st.plotly_chart(fig_race, use_container_width=True)

            # Log Table
            with st.expander("📜 View Detailed Game Log"):
                st.dataframe(df, use_container_width=True)
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
from logic import MultiplayerGame, run_monte_carlo_simulation, build_transition_matrix, ladders, snakes
import plotly.express as px
from st_aggrid import AgGrid, GridOptionsBuilder

# ---------- Page Config ----------
st.set_page_config(
    layout="wide",
    page_title="Snake & Ladder Project 🎲",
    page_icon="🎮"
)

# ---------- Custom CSS ----------
st.markdown("""
<style>
    .stApp {background-color: #f0f8ff;}
    .stHeader, h1, h2 {color: #2c3e50;}
    .stButton>button {background-color:#2c3e50; color:white; height:3em; width:100%; border-radius:10px;}
    .highlight {background-color: #ffff99;}
</style>
""", unsafe_allow_html=True)

# ---------- Title ----------
st.title("🎲 Snake & Ladder Project - Step-by-Step Live Game")

# ---------- Sidebar ----------
st.sidebar.header("Settings")
sims = st.sidebar.slider("Number of simulations", 100, 2000, 500)
start_player = st.sidebar.radio("Who starts?", ["Player A", "Player B"])

# ---------- Tabs ----------
tab1, tab2, tab3 = st.tabs(["📊 Simulation", "🧮 Transition Matrix", "🎮 Live Game"])

# ---------- Session State ----------
if 'game' not in st.session_state:
    st.session_state.game = None
if 'game_history' not in st.session_state:
    st.session_state.game_history = pd.DataFrame()
if 'stats' not in st.session_state:
    st.session_state.stats = {'A': {'Ladders':0,'Snakes':0,'Turns':0},
                              'B': {'Ladders':0,'Snakes':0,'Turns':0}}

# ---------- Tab 1: Simulation ----------
with tab1:
    st.header("Win Distribution")
    if st.button("Run Simulation", key="sim"):
        with st.spinner('Calculating...'):
            turns, winners = run_monte_carlo_simulation(sims)
            df_wins = pd.DataFrame(winners, columns=['Winner'])
            df_wins_count = df_wins['Winner'].value_counts().reset_index()
            df_wins_count.columns = ['Winner', 'Count']
            color_map = {'A':'blue','B':'red'}
            fig_wins = px.bar(df_wins_count, x='Winner', y='Count', color='Winner',
                              color_discrete_map=color_map, title="Number of Wins per Player")
            st.plotly_chart(fig_wins)
            st.success(f"Average turns to win: {np.mean(turns):.1f}")

# ---------- Tab 2: Transition Matrix ----------
with tab2:
    st.header("Markov Transition Matrix Analysis")
    P = build_transition_matrix(ladders, snakes)
    fig = px.imshow(P, labels=dict(x="To", y="From", color="Probability"), color_continuous_scale='Viridis')
    st.plotly_chart(fig)

# ---------- Tab 3: Live Game ----------
with tab3:
    st.header("Step-by-Step Live Game")

    # زر بدء لعبة جديدة
    if st.button("Start New Game", key="new_game"):
        st.session_state.game = MultiplayerGame(ladders, snakes,
                                                first_mover='A' if start_player=="Player A" else 'B')
        st.session_state.game_history = pd.DataFrame()
        st.session_state.stats = {'A': {'Ladders':0,'Snakes':0,'Turns':0},
                                  'B': {'Ladders':0,'Snakes':0,'Turns':0}}

    # زر حركة واحدة
    if st.session_state.game is not None:
        if st.button("Play Next Turn", key="next_turn"):
            move = st.session_state.game.play_one_turn()

            # احصائيات Ladder/ Snake
            player = move['Player']
            if move['To'] > move['From']:  # Ladder
                if move['To'] in ladders.values():
                    st.session_state.stats[player]['Ladders'] += 1
            elif move['To'] < move['From']: # Snake
                if move['To'] in snakes.values():
                    st.session_state.stats[player]['Snakes'] += 1
            st.session_state.stats[player]['Turns'] += 1

            # تحديث الجدول
            st.session_state.game_history = pd.DataFrame(st.session_state.game.get_game_stats())

        df = st.session_state.game_history
        if not df.empty:
            # تلوين الصف الأخير
            last_index = df.index[-1]
            gb = GridOptionsBuilder.from_dataframe(df)
            gb.configure_default_column(filterable=True, sortable=True)
            gb.configure_grid_options(domLayout='normal',
                                      getRowStyle={'function':'function(params) {if(params.node.rowIndex == '+str(last_index)+') return {backgroundColor:"#ffff99"};}'})
            grid_options = gb.build()
            AgGrid(df, gridOptions=grid_options, height=400, theme='light')

            # رسم خط تقدم اللاعب
            fig_progress = px.line(df, x='Turn', y='To', color='Player', markers=True,
                                   title="Player Progress Over Game")
            st.plotly_chart(fig_progress)

            # عرض الإحصائيات الجانبية
            stats_df = pd.DataFrame(st.session_state.stats).T
            stats_df['Avg Turns'] = stats_df['Turns'] / stats_df['Turns'].max() * 100
            st.subheader("Player Statistics")
            st.dataframe(stats_df)

            # تحقق من الفائز
            winner = st.session_state.game.get_winner()
            if winner:
                st.success(f"🏆 Winner: {winner}")
                st.session_state.game = None  # تنتهي اللعبة
```
